```python
#Cantidad de personas desempleadas
# Usamos el DataFrame original, que está en la variable df_d en la celda 6
import pandas as pd

df_desocupados = pd.read_csv('https://github.com/niconomist98/DataAnalyticsUQ/raw/refs/heads/main/Datos/GEIH2025/No%20ocupados.CSV', encoding='latin', sep=';')
D = df_desocupados[df_desocupados['DSI'] == 1]['FEX_C18'].sum()
print('La cantidad de personas desempleadas es:', int(D))
#Tasa de desempleados
td = (D / pea)
print(f'La tasa de desempleo es: {td*100} %')
```

```python
#punto 2
# Hacemos un merge para tener el sexo de los desocupados
# Usa el DataFrame original de la celda 6, no el escalar de la celda 15
df_desocupados_gen = pd.merge(
    df_desocupados[df_desocupados['DSI'] == 1],
    df_cgsse2[['DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'P3271']],
    on=['DIRECTORIO', 'SECUENCIA_P', 'ORDEN'],
    how='left'
)
desocupados_sexo = df_desocupados_gen.groupby(['DPTO', 'P3271'])['FEX_C18'].sum().reset_index().rename(columns={'FEX_C18': 'Desocupados'})

# Ocupados también con sexo
df_ocupados_gen = pd.merge(
    df_o,
    df_cgsse2[['DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'P3271']],
    on=['DIRECTORIO', 'SECUENCIA_P', 'ORDEN'],
    how='left'
)
ocupados_sexo = df_ocupados_gen.groupby(['DPTO', 'P3271'])['FEX_C18'].sum().reset_index().rename(columns={'FEX_C18': 'Ocupados'})

# Agrupamos desocupados y ocupados para calcular la PEA
pea_sexo = pd.merge(ocupados_sexo, desocupados_sexo, on=['DPTO', 'P3271'], how='outer').fillna(0)
pea_sexo['PEA'] = pea_sexo['Ocupados'] + pea_sexo['Desocupados']

# Calculamos la Tasa de Desempleo, manejando el caso de PEA=0
pea_sexo['TD'] = pea_sexo['Desocupados'] / pea_sexo['PEA'] * 100

# Reemplazamos sexo por etiquetas
pea_sexo['sexo'] = pea_sexo['P3271'].map({1: 'Hombres', 2: 'Mujeres'})

print(pea_sexo[['DPTO', 'sexo', 'TD']].head(20))
```

```python
# mayor y menor tasa de desempleo por sexo
max_td = pea_sexo.loc[pea_sexo.groupby('sexo')['TD'].idxmax()]
min_td = pea_sexo.loc[pea_sexo.groupby('sexo')['TD'].idxmin()]

print("departamento con mayor tasa de desempleo por sexo:")
print(max_td[['DPTO', 'sexo', 'TD']])

print("\ndepartamento con menor tasa de desempleo por sexo:")
print(min_td[['DPTO', 'sexo', 'TD']])
```

```python
#punto 3
# hacer merge para traer el genero (P3271: 1= hombre, 2= mujer)
df_ocupados_gen = pd.merge(df_desocupados, df_cgsse2[['DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'P3271']], on=['DIRECTORIO', 'SECUENCIA_P', 'ORDEN'], how='left')

#calcular hombres desocupados
hombres_desocupados = df_ocupados_gen[df_ocupados_gen['P3271'] == 1]['FEX_C18'].sum()
print('La cantidad de hombres desempleados es:', int(hombres_desocupados))

#Calcular mujeres desocupadas
mujeres_desocupadas = df_ocupados_gen[df_ocupados_gen['P3271'] == 2]['FEX_C18'].sum()
print(f'La cantidad de mujeres desempleadas es: {int(mujeres_desocupadas)}')
```

```python
# Renombra 'P3271_y' a 'P3271' en el DataFrame de ocupados
df_ocupados_gen = df_ocupados_gen.rename(columns={'P3271_y': 'P3271'})
```

```python
# punto 3.2
# Clasificamos a la población por edad en cada departamento para encontrar el grupo con mayor y menor desempleo. Para este estudio, no se consideró a las personas menores de 18 ni a las mayores de 59 años.
def clasificar_edaad(edad):
    if 18 <= edad < 24:
        return '18-24'
    elif 24 <= edad <= 30:
        return '24-30'
    elif 31 <= edad <= 40:
        return '31-40'
    elif 41 <= edad <= 50:
        return '41-50'
    elif edad >= 60:
        return '60+'
    else:
        return 'Otros'
```

```python
# armamos la PEA por departamento y grupo de edad
# agrupamos 
# creacion de columnas para grupo de edad en df_cgsse2


df_cgsse2['grupo_edad'] = df_cgsse2['P6040'].apply(clasificar_edaad)
# Asegúrate de haber ejecutado las celdas donde se definen df_cgsse2 y clasificar_edaad antes de ejecutar esta celda.

# Hacemos un merge para tener el grupo de edad de los desocupados
df_desocupados_gen = pd.merge(
    df_desocupados[df_desocupados['DSI'] == 1],
    df_cgsse2[['DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'grupo_edad']],
    on=['DIRECTORIO', 'SECUENCIA_P', 'ORDEN'],
    how='left'
)
desocupados_edad = df_desocupados_gen.groupby(['DPTO', 'grupo_edad'])['FEX_C18'].sum().reset_index().rename(columns={'FEX_C18': 'Desocupados'})

# Ocupados también con grupo de edad
df_ocupados_gen = pd.merge(
    df_o,
    df_cgsse2[['DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'grupo_edad']],
    on=['DIRECTORIO', 'SECUENCIA_P', 'ORDEN'],
    how='left'
)
ocupados_edad = df_ocupados_gen.groupby(['DPTO', 'grupo_edad'])['FEX_C18'].sum().reset_index().rename(columns={'FEX_C18': 'Ocupados'})

# tasa de desocupados y PEA
pea_edad = pd.merge(ocupados_edad, desocupados_edad, on=['DPTO', 'grupo_edad'], how='outer').fillna(0)
pea_edad['PEA'] = pea_edad['Ocupados'] + pea_edad['Desocupados']
pea_edad['TD'] = pea_edad['Desocupados'] / pea_edad['PEA'] * 100
```

```python
pea_edad = pea_edad[pea_edad['grupo_edad'].isin(['18-24', '24-30', '31-40', '41-50', '60+'])]

# menor y mayor tasa de desempleo por departamento
menor_tasa = pea_edad.loc[pea_edad.groupby('DPTO')["TD"].idxmin()]
mayor_tasa = pea_edad.loc[pea_edad.groupby('DPTO')["TD"].idxmax()]

# hacemos print
print("departamento con mayor tasa de desempleo por grupo de edad:")
print(mayor_tasa[['DPTO', 'grupo_edad', 'TD']])
```

```python
#print menor tasa de desempleo por departamento
print("\ndepartamento con menor tasa de desempleo por departamento:")
print(menor_tasa[['DPTO', 'grupo_edad', 'TD']])
```
